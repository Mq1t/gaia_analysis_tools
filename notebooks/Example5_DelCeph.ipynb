```python
#in this example we will observe the epoch photometry of a Cepheid Variable Star: Gaia DR3 2200153454733285248
#Importing the necessary libraries
import sys, os
sys.path.append(os.path.abspath(".."))
import gaia_analysis_tools.gaia_input as gi
import gaia_analysis_tools.epoch_photometry as ephot
```

```python
#Logging into the Gaia Archive makes it easier to access the data for datalinks (epoch photometry) but it is not necessary.
#You can create an account at https://www.cosmos.esa.int/web/gaia-users/register to log in. If you do not wish to log in, press n when prompted
gi.gaia_login_prompt()
```

```python
#In this example we will show how to cross match ID for Del Cep, using its HD ID: HD 213306.
#By simply inputting the HD ID into the query_by_datalink function, we can retrieve the Gaia DR3 source ID for Del Cepheid as well as its data
df_datalink =  gi.query_by_datalink("HD 213306")
```

```python
#Then with the datalink data, we can plot the epoch photometry of Del Cepheid
ephot.lightcurve(df_datalink, title="Flux vs Time of the Star Example 5")
ephot.lomb_scargle(df_datalink, plot = True, title="Lomb-Scargle Periodogram of the Star Example 5")
ephot.pdm(df_datalink, plot = True, title="PDM Plot of the Star Example 5")
```
